5. Generate a synthetic dataset with correlated features. Visualize the correlation matrix of a dataset using a
heatmap.

```
# Generate synthetic dataset
np.random.seed(42)  # for reproducibility
size = 200

# Create correlated features
x1 = np.random.normal(0, 1, size)
x2 = 2 * x1 + np.random.normal(0, 0.5, size)   # strongly correlated with x1
x3 = -1 * x1 + np.random.normal(0, 0.5, size)  # negatively correlated with x1
x4 = np.random.normal(0, 1, size)              # weak correlation

# Build DataFrame
data = pd.DataFrame({
    'Feature 1': x1,
    'Feature 2': x2,
    'Feature 3': x3,
    'Feature 4': x4
})

# Compute correlation matrix
corr_matrix = data.corr()

# Visualize with heatmap
plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix Heatmap")
plt.show()
```

3. Using the sales data, generate a heatmap to visualize the variation in sales across different months and
days.

```
# Seed for reproducibility
np.random.seed(20)

# Dataset
data = {
    'Month': np.random.choice(['Jan', 'Feb', 'Mar', 'Apr', 'May'], 100),
    'Day': np.random.choice(range(1, 31), 100),
    'Sales': np.random.randint(1000, 5000, 100)
}
df = pd.DataFrame(data)

# Pivot table for heatmap (average sales per day per month)
pivot_df = df.pivot_table(
    index="Day",
    columns="Month",
    values="Sales",
    aggfunc="mean"
)

# Create heatmap
fig = px.imshow(
    pivot_df,
    labels=dict(x="Month", y="Day", color="Average Sales"),
    title="Sales Variation Across Months and Days",
    aspect="auto",
    color_continuous_scale="Viridis"
)

fig.show()
```

 2. Using the given x and y data, generate a 3D surface plot to visualize the function z=\sin \left( \sqrt{x^2+y^2}\right)



```
import plotly.graph_objects as go

# Define grid
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x, y)

# Function
z = np.sin(np.sqrt(x**2 + y**2))

# Create surface plot
fig = go.Figure(data=[go.Surface(z=z, x=x, y=y, colorscale="Viridis")])

# Layout settings
fig.update_layout(
    title="3D Surface Plot of z = sin(sqrt(x² + y²))",
    scene=dict(
        xaxis_title="X Axis",
        yaxis_title="Y Axis",
        zaxis_title="Z Axis"
    ),
    width=800,
    height=600
)

fig.show()
```

2.Create a Bokeh scatter plot using randomly generated x and y values. Use different sizes and colors for the
markers based on the 'sizes' and 'colors' columns.

```
from bokeh.plotting import figure, show
from bokeh.io import output_notebook

# Display plots inline (use output_file("scatter.html") if you want a standalone HTML file)
output_notebook()

# Generate random data
np.random.seed(42)
data = {
    'x': np.random.rand(50) * 10,       # random x values
    'y': np.random.rand(50) * 10,       # random y values
    'sizes': np.random.randint(10, 40, 50),   # random sizes
    'colors': np.random.choice(['red', 'blue', 'green', 'orange', 'purple'], 50)  # random colors
}
df = pd.DataFrame(data)

# Create figure
p = figure(
    title="Bokeh Scatter Plot with Sizes and Colors",
    x_axis_label="X",
    y_axis_label="Y",
    width=700,
    height=400
)

# Add scatter points
p.scatter(
    x=df['x'],
    y=df['y'],
    size=df['sizes'],
    color=df['colors'],
    alpha=0.7
)

# Show plot
show(p)
```

3. Generate a Bokeh bar chart representing the counts of different fruits using the following dataset.

```
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
from bokeh.models import ColumnDataSource

# Display plots inline (use output_file("fruits.html") if you want a standalone HTML file)
output_notebook()

# Dataset
fruits = ['Apples', 'Oranges', 'Bananas', 'Pears']
counts = [20, 25, 30, 35]

# Convert to ColumnDataSource
source = ColumnDataSource(data=dict(fruits=fruits, counts=counts))

# Create figure
p = figure(
    x_range=fruits,
    title="Fruit Counts",
    x_axis_label="Fruit",
    y_axis_label="Count",
    width=700,
    height=400
)

# Add bars
p.vbar(
    x='fruits',
    top='counts',
    width=0.6,
    source=source,
    color="orange"
)

# Show plot
show(p)
```

4. Create a Bokeh histogram to visualize the distribution of the given data.

```

# Display plots inline (use output_file("histogram.html") if you want a standalone HTML file)
output_notebook()

# Generate data
data_hist = np.random.randn(1000)

# Compute histogram
hist, edges = np.histogram(data_hist, bins=30)

# Create figure
p = figure(
    title="Histogram of Random Data",
    x_axis_label="Value",
    y_axis_label="Frequency",
    width=700,
    height=400
)

# Add histogram bars
p.quad(
    top=hist,
    bottom=0,
    left=edges[:-1],
    right=edges[1:],
    fill_color="navy",
    line_color="white",
    alpha=0.7
)

# Show plot
show(p)
```

5. Create a Bokeh heatmap using the provided dataset.

```
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource, LinearColorMapper, ColorBar
from bokeh.transform import transform
from bokeh.palettes import Viridis256
from bokeh.io import output_notebook

# Enable notebook output
output_notebook()

# Provided dataset
data_heatmap = np.random.rand(10, 10)

x = np.linspace(0, 1, 10)
y = np.linspace(0, 1, 10)

xx, yy = np.meshgrid(x, y)

# Flatten the arrays for Bokeh
source = ColumnDataSource(data=dict(
    x=xx.flatten(),
    y=yy.flatten(),
    value=data_heatmap.flatten()
))

# Create color mapper
color_mapper = LinearColorMapper(
    palette=Viridis256,
    low=data_heatmap.min(),
    high=data_heatmap.max()
)

# Create figure
p = figure(
    title="Bokeh Heatmap",
    x_axis_label="X Axis",
    y_axis_label="Y Axis",
    width=600,
    height=600,
    tools="hover,save,pan,box_zoom,reset,wheel_zoom"
)

# Draw heatmap rectangles
p.rect(
    x='x',
    y='y',
    width=0.1,
    height=0.1,
    source=source,
    fill_color=transform('value', color_mapper),
    line_color=None
)

# Add color bar
color_bar = ColorBar(
    color_mapper=color_mapper,
    location=(0, 0)
)

p.add_layout(color_bar, 'right')

# Show plot
show(p)
```

```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```
