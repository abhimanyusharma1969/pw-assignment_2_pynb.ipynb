Write a code to perform basic string compression using the counts of repeated characters

```python
def compress_string(s):
    # Edge case: if the string is empty or just one character, return it as is
    if not s or len(s) == 1:
        return s

    # Initialize the compressed string and count variable
    compressed = []
    count = 1

    # Loop through the string starting from the second character
    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1  # Increase the count if the current character is the same as the previous one
        else:
            compressed.append(s[i - 1] + str(count))  # Append the previous character and its count
            count = 1  # Reset count for the new character

    # Append the last character and its count
    compressed.append(s[-1] + str(count))

    # Join the list into a single string
    compressed_string = ''.join(compressed)

    # Return the original string if the compressed version is not shorter
    return compressed_string if len(compressed_string) < len(s) else s

# Take input from the user
input_string = input("Enter a string to compress: ")

# Perform string compression
compressed_result = compress_string(input_string)

# Display the result
print(f"Compressed string: {compressed_result}")
```

Write a code to reverse a list in-place without using any built-in reverse functions

```python
def reverse_list_in_place(lst):
    start = 0
    end = len(lst) - 1

    while start < end:
        # Swap elements at start and end indices
        lst[start], lst[end] = lst[end], lst[start]

        # Move the pointers towards the center
        start += 1
        end -= 1

# Taking input from the user
input_list = input("Enter a list of elements separated by spaces: ").split()

# Calling the function to reverse the list
reverse_list_in_place(input_list)

# Printing the reversed list
print("Reversed list:", input_list)
```

Write a code to merge two sorted lists into a single sorted list

```python
def merge_sorted_lists(list1, list2):
    # Initialize two pointers and an empty list to store the result
    i, j = 0, 0
    merged_list = []

    # Merge the lists while both have elements
    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            merged_list.append(list1[i])
            i += 1
        else:
            merged_list.append(list2[j])
            j += 1

    # Add any remaining elements from list1 or list2
    merged_list.extend(list1[i:])
    merged_list.extend(list2[j:])

    return merged_list

# Taking input from the user
list1 = list(map(int, input("Enter the first sorted list of elements separated by spaces: ").split()))
list2 = list(map(int, input("Enter the second sorted list of elements separated by spaces: ").split()))

# Merging the lists
merged_list = merge_sorted_lists(list1, list2)

# Printing the merged sorted list
print("Merged sorted list:", merged_list)
```

Write a code to shuffle a given list randomly without using any built-in shuffle functions

```python
import random

def shuffle_list(lst):
    # Go through each index in the list
    for i in range(len(lst)):
        # Select a random index from the current position to the end of the list
        random_index = random.randint(i, len(lst) - 1)

        # Swap the current element with the randomly selected element
        lst[i], lst[random_index] = lst[random_index], lst[i]

# Taking input from the user
input_list = list(map(int, input("Enter a list of elements separated by spaces: ").split()))

# Shuffling the list
shuffle_list(input_list)

# Printing the shuffled list
print("Shuffled list:", input_list)
```

common keys, the values should be added together

```python
def merge_dictionaries(dict1, dict2):
    # Create a new dictionary that is a copy of the first dictionary
    merged_dict = dict1.copy()

    # Loop through each key-value pair in the second dictionary
    for key, value in dict2.items():
        # If the key is already in the first dictionary, add the values together
        if key in merged_dict:
            merged_dict[key] += value
        else:
            # If the key is not in the first dictionary, add the key-value pair
            merged_dict[key] = value

    return merged_dict

# Taking input from the user for the first dictionary
input_dict1 = input("Enter the first dictionary (key:value pairs separated by commas, and each pair by a space): ")
input_dict1 = dict(item.split(":") for item in input_dict1.split())

# Taking input from the user for the second dictionary
input_dict2 = input("Enter the second dictionary (key:value pairs separated by commas, and each pair by a space): ")
input_dict2 = dict(item.split(":") for item in input_dict2.split())

# Converting string values to integers for correct addition
input_dict1 = {k: int(v) for k, v in input_dict1.items()}
input_dict2 = {k: int(v) for k, v in input_dict2.items()}

# Merging the dictionaries
result = merge_dictionaries(input_dict1, input_dict2)

# Printing the merged dictionary
print("Merged dictionary:", result)
```

Write a code that inverts a dictionary, swapping keys and values. Ensure that the inverted dictionary
correctly handles cases where multiple keys have the same value by storing the keys as a list in the
inverted dictionary.

```python
def invert_dictionary(original_dict):
    inverted = {}
    for key, value in original_dict.items():
        # If the value is already a key in the inverted dictionary, append the key
        if value in inverted:
            inverted[value].append(key)
        else:
            # Otherwise, create a new entry with the key in a list
            inverted[value] = [key]
    return inverted

# Taking input from the user
import ast

user_input = input("Enter a dictionary (e.g., {'a': 1, 'b': 2, 'c': 1}): ")
input_dict = ast.literal_eval(user_input)

# Inverting the dictionary
inverted_dict = invert_dictionary(input_dict)

# Printing the result
print("Inverted dictionary:", inverted_dict)
```
